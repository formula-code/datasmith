## Helper Functions

```python
def fetch_pr(owner: str, repo: str, num: int) -> dict[str, Any]:
    """Fetch PR metadata from GitHub API."""
    pr = _get_github_metadata(endpoint=f"/repos/{owner}/{repo}/pulls/{num}")
    assert isinstance(pr, dict), "PR metadata should be a dict"
    return pr


def build_pr_dict(owner: str, repo: str, num: int) -> dict[str, Any]:
    """Build a PR dictionary suitable for ReportBuilder."""
    pr = fetch_pr(owner, repo, num)
    return {"pr_" + k: v for k, v in pr.items()}


def analyze_pr(pr_url: str, enable_llm: bool = False, summarize_llm: bool = False):
    """Analyze a PR and display results."""
    owner, repo, num_s = _parse_pr_url(pr_url)
    num = int(num_s)
    # Fetch PR data
    pr_dict = build_pr_dict(owner, repo, num)
    rb = ReportBuilder(
        enable_llm_backends=enable_llm,
        summarize_llm=summarize_llm,
        anonymize_output=False
    )
    result = rb.build(pr_dict)
    print("📋 PR Metadata:")
    print(f"  Title: {pr_dict['pr_title']}")
    print(f"  Merged at: {pr_dict['pr_merged_at']}")
    print(f"  Body length: {len(pr_dict['pr_body'] or '')} chars")
    print()

    # Issue extraction
    raw_issues = result.all_data.get("raw_issue_data", [])
    print(f"📌 Linked Issues Found: {len(raw_issues)}")
    if raw_issues:
        for issue in raw_issues:
            print(f"  - Issue #{issue.get('number')}: {issue.get('title', 'N/A')[:60]}...")
    print()

    # Problem statement
    problem = result.all_data.get("problem_statement", "")
    print(f"📝 Problem Statement ({len(problem)} chars):")
    if problem:
        print(f"  {problem[:200]}...")
    else:
        print("  [None]")
    print()

    # Solution overview
    solution = result.all_data.get("solution_overview", "")
    print(f"💡 Solution Overview ({len(solution)} chars):")
    if solution:
        print(f"  {solution[:200]}...")
    else:
        print("  [None]")
    print()

    # Report status
    report_status = "✅ Valid" if result.report_md != "NOT_A_VALID_PR" else "❌ Rejected"
    print(f"Status: {report_status}")
    print()
    return pr_dict, rb, result
```

## Test Case 1: Astropy PR #13498

Expected: Should recover issue #13479

```python
astropy_pr, astropy_result_builder, astropy_result = analyze_pr("https://github.com/astropy/astropy/pull/13498")
```

```python
from jinja2 import Template

astropy_pr['pr_base']['repo']

# make a jinja2 template string to display the PR metadata nicely
repo_description_template_str = """
This is {{ repo_name }}. {{ repo_name }} is written primarily in {{ repo_language }} and is described as a "{{ repo_description }}", with topics including {{ repo_topics }}.
""".strip()
repo_description_template = Template(repo_description_template_str)
repo_desc = repo_description_template.render(
    repo_name=astropy_pr['pr_base']['repo']['name'],
    repo_description=astropy_pr['pr_base']['repo']['description'],
    repo_topics=", ".join(astropy_pr['pr_base']['repo']['topics']),
    repo_language=astropy_pr['pr_base']['repo']['language']
)
print(repo_desc)
print(astropy_pr["pr_body"])
```

## Test Case 2: Qiskit PR #10651

Expected: No linked issues, but should have good problem/solution segmentation with >90% LCS match

```python
qiskit_10651_result, qiskit_10651_pr = analyze_pr("https://github.com/Qiskit/qiskit/pull/10651")
```

```python
# Check extractiveness of problem/solution
body = qiskit_10651_pr["pr_body"]
extractor = ProblemExtractor(validate_lcs=True)
extraction, validation = extractor.extract_problem(message=body, related_issues="")

print("Problem/Solution Extraction Analysis:")
print("\nProblem Statement:")
print(extraction.problem_statement)
print("\nSolution Overview:")
print(extraction.solution_overview)
print("\nValidation Metrics:")
pprint(validation)
```

## Summary Comparison

Let's create a summary table comparing all PRs

```python
import pandas as pd

results = [
    ("astropy/astropy#13498", astropy_result),
    ("Qiskit/qiskit#10651", qiskit_10651_result),
    ("Qiskit/qiskit#12869", qiskit_12869_result),
    ("apache/arrow#36738", arrow_result),
    ("apache/arrow-adbc#2697", arrow_adbc_result),
]

summary_data = []
for name, result in results:
    summary_data.append({
        "PR": name,
        "Valid": result.report_md != "NOT_A_VALID_PR",
        "# Issues": len(result.all_data.get("raw_issue_data", [])),
        "Has Problem": bool(result.all_data.get("problem_statement", "").strip()),
        "Has Solution": bool(result.all_data.get("solution_overview", "").strip()),
        "Problem Len": len(result.all_data.get("problem_statement", "")),
        "Solution Len": len(result.all_data.get("solution_overview", "")),
    })

df = pd.DataFrame(summary_data)
display(df)
```
