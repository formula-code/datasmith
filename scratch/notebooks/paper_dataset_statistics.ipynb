```python
descriptions = {}


def get_repo_description(repo_name: str) -> str:
    try:
        metadata = _get_github_metadata(f"repos/{repo_name}")
        description = metadata.get("description", "")
        descriptions[repo_name] = description
        return description
    except Exception:
        return ""


repos["description"] = repos["repo_name"].apply(get_repo_description)
```

```python
# save descriptions as json
with open("scratch/artifacts/pipeflush/repo_descriptions.json", "w") as f:
    json.dump(descriptions, f, indent=4)
```
